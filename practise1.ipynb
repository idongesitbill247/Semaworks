```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
hr_dataset = pd.read_excel(r"C:\Users\PC\Downloads\HR_Analytics_Pandas_Practice_Dataset.xlsx", sheet_name=['Employee_Master', 'Monthly_Activity'])
```

```python
employee_master = hr_dataset['Employee_Master']
```

```python
monthly_activity = hr_dataset['Monthly_Activity']
```

```python
employee_master
```

```python
type(employee_master)
```

```python
employee_master.info()
```

```python
employee_master.isna().sum()
```

```python
employee_master.duplicated().sum()
```

```python
employee_master = employee_master.drop_duplicates()
```

```python
employee_master.duplicated().sum()
```

```python
type(employee_master)
```

```python
employee_master.columns = (
    employee_master.columns
    .str.strip()
    .str.lower()
)
```

```python
employee_master['monthly_salary_ngn'] = pd.to_numeric(employee_master.monthly_salary_ngn, errors= 'coerce')
```

```python
employee_master.info()
```

```python
employee_master['department'] = employee_master['department'].str.strip().str.title()
```

```python
employee_master['annual_salary'] = employee_master['monthly_salary_ngn'] * 12
employee_master
```

```python
employee_master['training_courses'] = employee_master['training_courses'].str.split(',')
```

```python
em_exploded = employee_master.explode('training_courses')
em_exploded
```

```python
em_exploded['training_courses_clean'] = (
   em_exploded['training_courses'].astype(str)
   .str.strip()
   .str.strip("[]'\"")
   .str.strip()
   .str.title()
)
```

```python
employee_master['employment_status'] = employee_master['employment_status'].str.strip().str.title()
```

## Monthly activity


```python
monthly_activity
```

```python
type(monthly_activity)
```

```python
monthly_activity.info()
```

```python
monthly_activity.isna().sum()
```

```python
monthly_activity.duplicated().sum()
```

```python
monthly_activity.columns = (
    monthly_activity.columns
    .str.strip()
    .str.lower()
)
```

```python
monthly_activity['month_num'] = monthly_activity['month'].dt.month
monthly_activity['months'] = monthly_activity['month'].dt.strftime('%b')
```

## Merged & Analysed Data


```python
employee_record_and_activities = employee_master.merge(monthly_activity, on= 'employee_id')
```

```python
employee_record_and_activities
```

```python
type(employee_record_and_activities)
```

```python
employee_record_and_activities.info()
```

```python
headcount_by_dept = employee_record_and_activities.pivot_table(
    values= 'employee_id', index= 'department', aggfunc='nunique'
    ).sort_values(by= 'employee_id', ascending= False)
headcount_by_dept
```

```python
employment_type_by_workforce = employee_record_and_activities.pivot_table(
    values= 'employee_id', index= 'employment_type', aggfunc='nunique'
    ).sort_values(by= 'employee_id', ascending= False)
employment_type_by_workforce
```

```python
avg_salary_by_dept = employee_master.pivot_table(
    values= 'monthly_salary_ngn', index= 'department', aggfunc= 'mean'
    ).sort_values(by= 'monthly_salary_ngn', ascending= False)
avg_salary_by_dept
```

```python
salary_by_position = employee_master.pivot_table(
    values= 'monthly_salary_ngn',
    index= 'job_title',
    aggfunc= 'mean'
    
).sort_values(by= 'monthly_salary_ngn', ascending=False)
salary_by_position
```

```python
workforce_status = employee_master.pivot_table(
    values= 'employee_id',
    index= 'employment_status',
    aggfunc= 'count'
).sort_values(by= 'employee_id', ascending= False)
workforce_status
```

```python
employee_record_and_activities['perf_band'] = pd.cut(
    employee_record_and_activities['productivity_score'],
    bins= [0, 55, 75, 100],
    labels= ['Low', 'Medium', 'High']
)

dept_performance = employee_record_and_activities.pivot_table(
    values= 'employee_id',
    index= 'department',
    columns= 'perf_band',
    aggfunc='count',
    fill_value= 0
)
dept_performance
```

```python
satisfaction_rate = employee_record_and_activities.pivot_table(
    values= 'monthly_satisfaction_score',
    index= 'department',
    aggfunc= 'mean'
)
satisfaction_rate
```

```python
trending_training = em_exploded.pivot_table(
    values= 'employee_id',
    index= 'training_courses_clean',
    aggfunc= 'count'
)
trending_training
```

```python
month_active_trend = employee_record_and_activities.pivot_table(
    values= 'productivity_score',
    index= ['month_num', 'months'],
    aggfunc= 'mean'
)
month_active_trend
```

```python
employee_record_and_activities.pivot_table(
    values= 'productivity_score',
    index= ['month_num', 'months'],
    columns= 'department',
    aggfunc= 'mean'
)
```

## Charts

```python
headcount_by_dept.plot(kind= 'bar', color= 'teal', figsize= (6,4), legend= False)
plt.ticklabel_format(style='plain', axis='y')
plt.title('Headcount by Department', fontweight= 'bold')
plt.ylabel('')
plt.yticks([])
plt.xlabel('Dept')
plt.xticks(rotation=45)
plt.show()
```

```python
colors = {
    'deep_teal': '#004D40',
    'forest': '#1B4332',
    'mint': '#52B788',
    'sage': '#74C69D',
    'pale': '#D8F3DC',
    'teal': '#00796B'
}

line_colors = ['#D8F3DC', '#004D40', '#1B4332', '#74C69D', '#52B788', '#40916C', '#95D5B2']

perf_color = [
    colors['pale'], # High
    colors['teal'], # Medium
    colors['deep_teal'], # low
]


fig, axes = plt.subplots(2, 3, figsize= (22, 16))
axes = axes.flatten()

# Dahbosrd title
fig.suptitle('EMPLOYEE PERFORMANCE DASHBOARD', fontsize= 22, fontweight= 'bold', color=colors['deep_teal'], y=1.02)

# 1 Headcount by dept

headcount_by_dept.plot(kind= 'bar', ax= axes[0], color=colors['deep_teal'], legend= False)
axes[0].set_title('Headcount by Department', fontsize= 12, fontweight= 'bold')
axes[0].set_ylabel('')
axes[0].set_xlabel('NO. of Employees')
axes[0].tick_params(axis= 'x', rotation=45)

# 2 Employment type
employment_type_by_workforce.plot(kind= 'barh', ax= axes[1], color=colors['forest'], legend= False)
axes[1].set_title('Workforce',fontsize= 12, fontweight= 'bold')

# 3 Salary by position

axes[2].hlines(y=salary_by_position.index, xmin= 0, xmax=salary_by_position.values, color=colors['teal'], linewidth=2)
axes[2].plot(salary_by_position.values, salary_by_position.index, 'o', color=colors['teal'], markersize=6)
axes[2].set_title('Average Salary by Position', fontsize= 12, fontweight= 'bold')
axes[2].set_xlabel('Average salary')
axes[2].set_ylabel('Position')
axes[2].tick_params(axis='y', labelsize= 8)
axes[2].grid(axis='x', linestyle= '--', alpha=0.4)

# 4 Department performance

dept_performance.plot(kind= 'bar', stacked= True, ax= axes[3], color= perf_color, edgecolor= 'white')
axes[3].set_title('Department Performance Band', fontsize= 12, fontweight= 'bold')
axes[3].set_xlabel('')
axes[3].set_ylabel('')
axes[3].tick_params(axis='x', labelrotation= 45)
axes[3].legend(title='Performance Band')

# 5 month_acitvity

month_active_trend.plot(kind= 'line', marker= 's', markersize= 8, linewidth= 3, ax=axes[4], color=colors['forest'])
axes[4].set_title('Monthly Activity Trend', fontsize= 12, fontweight= 'bold')
axes[4].set_xlabel('Month')
axes[4].set_ylabel('Avg Productivity Score')
axes[4].tick_params(axis='x', labelrotation= 0)
axes[4].grid(alpha=0.3)

# 6 trendin courses

trending_training.plot(kind= 'line', marker= 'o', markersize= 6, markerfacecolor= colors['deep_teal'], 
    linewidth= 3, ax= axes[5], color=colors['mint'])
axes[5].set_title('Trending Training Courses', fontsize= 12, fontweight= 'bold')
axes[5].set_xlabel('Courses')
axes[5].set_ylabel('Active trainees')
axes[5].tick_params(axis='x', labelrotation= 45)
axes[4].grid(alpha=0.3)



plt.tight_layout()
plt.show()
```
