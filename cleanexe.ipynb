## Creating more than two chart

```python
fig, axes = plt.subplots(1, 2, figsize= (12,5))
quarterly_revenue.plot(kind= 'line', marker= 's', color= 'green', linewidth= 3, ax=axes[0])
plt.ticklabel_format(style='plain', axis='y')
axes[0].set_title('Quarterly Revenue')
axes[0].set_xlabel('Quarter')
axes[0].set_ylabel('Revenue(ngn)')

revenue_by_branch.plot(kind= 'bar', color= 'teal', ax=axes[1])
plt.ticklabel_format(style='plain', axis='y')
axes[1].set_title('Revenue by Branch')
axes[1].set_ylabel('Revenue(ngn)')
axes[1].tick_params(axis='x', rotation= 45)

plt.tight_layout()
plt.show()
```
